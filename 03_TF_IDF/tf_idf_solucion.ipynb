# Actividad 2: TF-IDF

TF-IDF (*Term Frequency – Inverse Document Frequency*) es una métrica que pondera qué tan importante es una palabra dentro de un documento, en relación con el resto del corpus. Palabras muy comunes en todos los documentos (como *is*, *of*, *a*) reciben un peso bajo, mientras que palabras distintivas de un documento reciben un peso alto.

**Nota (tip del profesor):** en versiones antiguas de scikit-learn el método se llamaba `get_feature_names()`; en versiones recientes (>= 1.0) se reemplazó por `get_feature_names_out()`. Este notebook usa la versión actual instalada, por lo que se utiliza `get_feature_names_out()`. Si tu entorno da error con ese método, sustitúyelo por `get_feature_names()`.

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

# Definir el conjunto de documentos
documents = [
    "Data science is an interdisciplinary field",
    "Machine learning is a part of data science",
    "Data science involves statistics"
]

# Crear el vectorizador TF-IDF
vectorizer = TfidfVectorizer()

# Ajustar y transformar los documentos
tfidf_matrix = vectorizer.fit_transform(documents)

# Compatibilidad entre versiones de scikit-learn
try:
    feature_names = vectorizer.get_feature_names_out()
except AttributeError:
    # Versiones antiguas de scikit-learn (< 1.0)
    feature_names = vectorizer.get_feature_names()

# Crear un DataFrame para visualizar los resultados
df = pd.DataFrame(tfidf_matrix.toarray(), columns=feature_names)
print(df.round(3))
```

### Interpretación rápida
- La palabra **"science"** aparece en los 3 documentos, por lo que su peso IDF es bajo (no es muy distintiva).
- Palabras como **"interdisciplinary"**, **"statistics"** o **"machine"** solo aparecen en un documento, por lo que reciben un peso TF-IDF más alto: son más representativas de ese documento en particular.
